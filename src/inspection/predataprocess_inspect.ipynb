```python
def num_classes(df):
    class_list = df["subclass"].values

    o_count = len([x for x in class_list if x[0] == "O"])
    b_count = len([x for x in class_list if x[0] == "B"])
    a_count = len([x for x in class_list if x[0] == "A"])
    f_count = len([x for x in class_list if x[0] == "F"])
    g_count = len([x for x in class_list if x[0] == "G"])
    k_count = len([x for x in class_list if x[0] == "K"])
    m_count = len([x for x in class_list if x[0] == "M"])


    return {"O" : o_count, "B" : b_count, "A": a_count, "F": f_count,  "G": g_count, "K": k_count, "M" : m_count}

    
```

```python
print("LAMOST :")
print(num_classes(lamost_meta))
print("=======================================================================")
print("SDSS :")
print(num_classes(sdss_meta))
```

```python
import h5py
import numpy as np

# ── SDSS ──
sdss_file = h5py.File("/home/arbiter/projects/Survey-invariant-generalization/data/sdss_final_ready.h5")

sdss_median_snrs = []
for star in sdss_file:
    snr = sdss_file[star]["snr"][:]
    sdss_median_snrs.append(np.median(snr))

print(f"SDSS median SNR: {np.median(sdss_median_snrs):.2f}")
print(f"SDSS mean SNR:   {np.mean(sdss_median_snrs):.2f}")

# ── LAMOST ──
lamost_file = h5py.File("/home/arbiter/projects/Survey-invariant-generalization/data/lamost_combined.h5")

lamost_snrs = []
for star in lamost_file:
    snrg = lamost_file[star].attrs.get("SNRG", np.nan)
    snrr = lamost_file[star].attrs.get("SNRR", np.nan)
    mean_snr = np.nanmean([snrg, snrr])
    lamost_snrs.append(mean_snr)

print(f"LAMOST median SNR: {np.median(lamost_snrs):.2f}")
print(f"LAMOST mean SNR:   {np.mean(lamost_snrs):.2f}")
```

#### Inspecting the wavelength grid and the flux grid

```python
total_list = []
for i in lamost_file:
    empty_list = []
    wavelength_grid = lamost_file[i]["wavelength_grid"][:]
    start_value = lamost_file[i]["wavelength_grid"][0]
    end_value = lamost_file[i]["wavelength_grid"][-1]
    step_value = np.abs(lamost_file[i]["wavelength_grid"][2] - lamost_file[i]["wavelength_grid"][1])
    length_wave = len(wavelength_grid)
    length_flux = len(lamost_file[i]["flux"][:])
    empty_list.append(start_value)
    empty_list.append(end_value)
    empty_list.append(step_value)
    empty_list.append(length_wave)
    empty_list.append(length_flux)

    total_list.append(empty_list)

    
```

```python
from tqdm import tqdm
```

```python
def parse_SDSS(spectral_class):
    folder_path = f"/home/arbiter/projects/Survey-invariant-generalization/data/fits_files/SDSS_{spectral_class}_fits"

    dir_list = os.listdir(folder_path)

    df_star = pd.read_csv(f"{folder_path}/SDSS_{spectral_class}_meta_data.csv")
    
    total_dict = {}

    counter = 0

    for folder in tqdm(dir_list):
        if ".csv" not in folder:
            for file in os.listdir(f"{folder_path}/{folder}"):

                star_dict = {}
                with fits.open(f"{folder_path}/{folder}/{file}") as star_file:
                    primary_hdu = star_file[0]
                    primary_header = primary_hdu.header
                    

                    secondary_hdu = star_file[1]
                    secondary_data = secondary_hdu.data

                    plate_id = primary_header["PLATEID"]
                    mjd = primary_header["MJD"]
                    fiber_id = primary_header["FIBERID"]

                    flux = [row[0] for row in secondary_data]
                    wavelength_grid = [10**row[1] for row in secondary_data]

                    snr = [row[0] * np.sqrt(row[2]) for row in secondary_data]

                    df_needed = df_star.loc[(df_star["plate"] == plate_id) & (df_star["mjd"] == mjd) & (df_star["fiberid"] == fiber_id)]

                    metallicity = float(df_needed["elodieFeH"].iloc[0])
                    t_eff = float(df_needed["elodieTeff"].iloc[0])
                    log_g = float(df_needed["elodieLogG"].iloc[0])

                    stellar_class = str(df_needed["subclass"].iloc[0][0])

                star_dict["flux"] = flux
                star_dict["wavelength_grid"] = wavelength_grid
                star_dict["plate"] = plate_id
                star_dict["fiber"] = fiber_id
                star_dict["mjd"] = mjd
                star_dict["metallicity"] = metallicity
                star_dict["t_eff"] = t_eff
                star_dict["log_g"] = log_g
                star_dict["snr"] = snr
                star_dict["spectral_class"] = stellar_class

                total_dict[counter] = star_dict

                counter += 1

    
    return total_dict
    
```

```python
dict_A = parse_SDSS("G")
```

```python
def convert_float32_iterative(obj):
    stack = [obj]

    while stack:
        current = stack.pop()

        if isinstance(current, dict):
            for k, v in current.items():

                if isinstance(v, np.float32):
                    current[k] = float(v)

                elif isinstance(v, np.ndarray):
                    current[k] = v.astype(float)

                elif isinstance(v, (dict, list)):
                    stack.append(v)

        elif isinstance(current, list):
            for i, v in enumerate(current):

                if isinstance(v, np.float32):
                    current[i] = float(v)

                elif isinstance(v, np.ndarray):
                    current[i] = v.astype(float)

                elif isinstance(v, (dict, list)):
                    stack.append(v)

    return obj   # optional but convenient

dict_A = convert_float32_iterative(dict_A)
```

```python
with open("gsd.json", "w", encoding = "utf-8") as thukuna:
    json.dump(dict_A, thukuna, indent= 4)
```

```python
dict_K = parse_SDSS("K")
dict_K = convert_float32_iterative(dict_K)

with open("ksd.json", "w", encoding = "utf-8") as file:
    json.dump(dict_K, file, indent= 4)
```

```python
import numpy as np
import pandas as pd 

import matplotlib.pyplot as plt
import seaborn as sns

import sklearn
from sklearn.metrics import classification_report, accuracy_score
%matplotlib inline
```

```python
preds_reg = np.load("preds_reg.npy")
preds_cls = np.load("preds_cls.npy")

true_reg = np.load("true_reg.npy")
true_cls = np.load("true_cls.npy")
```

```python
# MAE for each parameter
mae_teff = np.mean(np.abs(preds_reg[:, 0] - true_reg[:, 0]))
mae_logg = np.mean(np.abs(preds_reg[:, 1] - true_reg[:, 1]))
mae_feh  = np.mean(np.abs(preds_reg[:, 2] - true_reg[:, 2]))
```

```python
cls_accuracy = accuracy_score(true_cls, preds_cls)
cls_accuracy
```

```python
print(f"MAE Teff:   {mae_teff:.4f}")
print(f"MAE logg:   {mae_logg:.4f}")
print(f"MAE [Fe/H]: {mae_feh:.4f}")
print(f"Accuracy:   {accuracy_score(true_cls, preds_cls):.4f}")
print()
print(classification_report(true_cls, preds_cls))
```

```python
reg_sigma = np.load("../../models/reg_sigma.npy")
```

```python
mae_teff_physical = mae_teff * reg_sigma[0]  # Kelvin
mae_logg_physical = mae_logg * reg_sigma[1]  # dex
mae_feh_physical  = mae_feh  * reg_sigma[2]  # dex

print(f"MAE Teff:   {mae_teff_physical:.2f} K")
print(f"MAE logg:   {mae_logg_physical:.4f} dex")
print(f"MAE [Fe/H]: {mae_feh_physical:.4f} dex")
```

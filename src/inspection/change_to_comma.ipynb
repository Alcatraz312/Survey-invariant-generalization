```python
def to_comma(dir):
    df = pd.read_csv(f"{dir}.csv", delim_whitespace= True)

    df.to_csv(f"{dir}_comma", index = False, quoting=csv.QUOTE_NONE, escapechar=" ")
```

```python
def to_comma_meta(directory):
    df = pd.read_csv(f"{directory}.csv")
    size = 0
    if len(df) == 4000:
        size = 4

    elif len(df) == 2000:
        size = 2

    elif len(df) <= 1000:
        size = 1

    for i in range(size):
        df_chunk = df.iloc[(i * 1000): (i + 1) * 1000][["plate","mjd","fiberid"]]
        
        df_chunk.to_csv(f"{dir}_comma_{i + 1}", index = False, quoting=csv.QUOTE_NONE, escapechar=" ")
```

```python
to_comma_meta("../../data/SDSS_B_meta_data")
```

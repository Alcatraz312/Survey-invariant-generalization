```python
import numpy as np
import pandas as pd 

import matplotlib.pyplot as plt
import seaborn as sns

from astroquery.sdss import SDSS 

%matplotlib inline
```

```python
def retrieve_data(spectra_class : str):

    spectra = np.load(f"../../data/spectra_{spectra_class}.npy", allow_pickle= True)    

    # if spectra_class == "F":
    #     num = 4290
    # else:
    #     num = 4285

    # running the sql query to get the fiberid, mjd, plate for the stars
    sql_query = f'''SELECT top 100000
        specobjid, ra, dec,plate, mjd, fiberid, subclass, z, elodieTeff, elodieLogG, elodieFeH
        FROM SpecObj
        WHERE class = 'STAR' AND zWarning = 0 
            AND subclass LIKE '{spectra_class}%'
            AND (snMedian_u + snMedian_g + snMedian_r + snMedian_i + snMedian_z)/5.0 > 10
            AND elodieTeff IS NOT NULL AND elodieLogG IS NOT NULL AND elodieFeH IS NOT NULL'''

    res = SDSS.query_sql(sql_query, timeout = 600)

    df = res.to_pandas()

    iter = 0
    dict_spectra = {}
    for array in spectra:
        empty_dict = {}
        uid = int(array[0])

        row = df.loc[df["specobjid"] == uid]

        if row.empty:
            continue   # skip this spectrum safely

        stellar_class = str(row["subclass"].iloc[0])
        teff = float(row["elodieTeff"].iloc[0])
        logg = float(row["elodieLogG"].iloc[0])
        feh  = float(row["elodieFeH"].iloc[0])


        empty_dict["Flux"] = array
        empty_dict["uid"] = uid
        empty_dict["Spectral class"] = stellar_class
        empty_dict["teff"] = teff
        empty_dict["logg"] = logg
        empty_dict["feh"] = feh

        dict_spectra[iter] = empty_dict
        iter+=1

    return dict_spectra
```

```python
def check_size(spectra_class : str):
    spectra = np.load(f"../../data/spectra_{spectra_class}.npy", allow_pickle= True)

    return len(spectra)
```

```python
size_A = check_size("A")
size_B = check_size("B")
size_F = check_size("F")
size_G = check_size("G")
size_K = check_size("K")
size_M = check_size("M")
size_O = check_size("O")
```
